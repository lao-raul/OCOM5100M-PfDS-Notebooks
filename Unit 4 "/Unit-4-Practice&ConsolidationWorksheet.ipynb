````python
import numpy as np

signal = np.array([0.5, -0.3, 1.2, -1.8, 0.9, -0.1, 2.1])

# Replace negatives with 0 using boolean indexing
signal[signal < 0] = 0
print(signal)

# Clip to [-1, 1] using np.where (restore original for this step)
signal = np.array([0.5, -0.3, 1.2, -1.8, 0.9, -0.1, 2.1])
clipped = np.where(signal < -1, -1, np.where(signal > 1, 1, signal))
print(clipped)
````

**Key reminders:**

- `shape` returns a tuple of dimensions; `size` is the total number of elements; `ndim` is the number of axes
- `np.random.seed()` is a function call, not an attribute assignment — always use parentheses
- When casting to a smaller integer type, NumPy truncates rather than rounds — `3.9` becomes `3`, not `4`
- Slices of a NumPy array return **views**, not copies — modifying a view modifies the original; use `.copy()` to avoid this
- Use `|` and `&` for element-wise boolean operations on arrays, never `or` and `and` — and always wrap each condition in parentheses
- `np.nanmean`, `np.nansum`, and `np.nanargmax` ignore `nan` values; the plain versions propagate `nan` through the result
- `axis=0` operates **down** the rows (collapses rows → one value per column); `axis=1` operates **across** columns (collapses columns → one value per row)
- When dividing a 2D array by a 1D row-of-sums, reshape the 1D array to `(-1, 1)` so broadcasting aligns correctly
- `reshape` cannot change the total number of elements; use `-1` to let NumPy infer one dimension automatically
- `ravel()` returns a view when possible; `flatten()` always returns a copy
- `ax.set_title`, `ax.set_xlabel`, and `ax.legend` are methods on the **Axes object** (`ax`), not on `plt`
- Lines only appear in the legend if they were plotted with a `label=` argument
