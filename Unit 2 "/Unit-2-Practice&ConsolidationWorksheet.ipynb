**Key reminders:**

- Conditions in `if/elif/else` are evaluated from top to bottom — the first `True` branch runs and the rest are skipped, even if later conditions are also `True`
- Python uses indentation to define code blocks — a statement in the wrong column changes the logic or causes an `IndentationError`
- Every instance method must have `self` as its first parameter — without it, Python raises a `TypeError` when the method is called
- A function without a `return` statement returns `None` — computing a value but not returning it discards it silently
- Use specific exception types in `except` clauses — catching the wrong type leaves the real error unhandled and the program crashes
- The `finally` block always runs, whether or not an exception occurred — use it for cleanup that must happen regardless
