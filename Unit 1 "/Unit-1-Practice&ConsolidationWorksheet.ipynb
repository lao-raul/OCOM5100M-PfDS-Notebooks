### Exercise 6.5 Adapt and extend

Filter passing students into a new dictionary using a for loop, then calculate the class average.

Expected output:
```
{'Alice': 85, 'Bob': 72, 'Charlie': 91, 'Eve': 78}
Class average: 78.8
```

````python
students = {"Alice": 85, "Bob": 72, "Charlie": 91, "Diana": 68, "Eve": 78}

passed = {}
for name, score in students.items():
    if score >= 70:
        passed[name] = score

total = 0
for score in students.values():
    total = total + score
average = total / len(students)

print(passed)
print(f"Class average: {average:.1f}")
````

`.items()` gives access to both the key and value on each iteration, allowing the condition to check the score while storing the name. `.values()` returns just the scores, which are added to a running total using a for loop. ```

Converting to a set removes duplicates instantly. Converting back to a list with `list()` gives you a sortable sequence. Note that `len(list_with_dups)` still returns 10 — the original list is unchanged.
